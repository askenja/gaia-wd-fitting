```python
import numpy as np 
```

```python
def a(p1,p2):
    return np.atan((p2[1] -p1[1])/(p2[0] -p1[0]))

def b(p,a):
    return p[1] - a*p[0]
```

```python

points = [[0.95,0.5],[2.75,0.67],[3.54,0.81],[5.21,0.91],[8.0,1.37]]
```

```python
alphas = [np.round(a(points[i], points[i+1]),3) for i in range(len(points)-1)]
```

```python
offsets = [np.round(b(points[i+1], alphas[i]),3) for i in range(len(points)-1)]
```

```python
print(points)
print(alphas)
print(offsets)
```
